```python
from google.colab.patches import cv2_imshow
from google.colab import drive
import numpy as np
import cv2
import matplotlib.pyplot as plt
drive.mount('/content/gdrive')
```

```python
img = cv2.imread('/content/gdrive/My Drive/IP/Images/gsy2.jpg')

print(img.shape)
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
#plt.imshow(gimg, cmap='gray')
cv2_imshow(gimg)
```

```python
outimg = cv2.blur(gimg, (7,7))
cv2_imshow(outimg)
```

```python
outimg = cv2.GaussianBlur(gimg, (7,7), 5)
cv2_imshow(outimg)
```

```python
outimg = cv2.medianBlur(gimg, 7)
cv2_imshow(outimg)
```

```python
outimages = gimg.copy()
#blurimage = cv2.GaussianBlur(outimages, (7,7), 5)
blurimage = cv2.medianBlur(outimages, 7)

x = 200 + 70
y = 200 + 65
size = 15

for h in range(y, y+size):
  for w in range(x, x+size):
    outimages[h][w] = blurimage[h][w]

cv2_imshow(outimages)
```

```python
img = cv2.imread('/content/gdrive/My Drive/IP/Images/lena256.jpg')

print(img.shape)
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
#plt.imshow(gimg, cmap='gray')
cv2_imshow(gimg)
```

```python
outimg = cv2.GaussianBlur(gimg, (7,7), 2)
cv2_imshow(outimg)
```

```python
edgimage = cv2.addWeighted(gimg, 1, outimg, -1, 0)
cv2_imshow(edgimage)
```

```python
unsharpimage = cv2.addWeighted(gimg, 1, edgimage, 2, 0)
cv2_imshow(unsharpimage )
```
